```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('train.txt' , sep = ';' , header = None , names = ['text' , 'emotion'])
```

```python
df
```

```python
df.isnull().sum()
```

```python
emotions = df['emotion'].unique()
emotions_numeric = {}
i = 0
for emo in emotions :
    emotions_numeric[emo] = i
    i += 1
df['emotion'] = df['emotion'].map(emotions_numeric)
```

```python
df
```

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
import string
```

```python
def remove_punc(txt):
    return txt.translate(str.maketrans('' , '' , string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

```python
def remove_num(txt):
    new = ""
    for i in txt :
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_num)
```

```python
def remove_emoji(txt):
    new = ""
    for i in txt :
        if not i.isascii():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_num)
```

```python
import nltk # to remove stopping words like is , am , the
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
stop_words
```

```python
len(stop_words)
```

```python
def remove_stop(txt):
    words = word_tokenize(txt)
    cleaned_text = []
    for i in words:
        if not i in stop_words:
            cleaned_text.append(i)
    return ' '.join(cleaned_text)
```

```python
df['text'] = df['text'].apply(remove_stop)
```

```python
df
```

```python
X = df['text']
y = df['emotion']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.30 , random_state = 42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer , CountVectorizer
```

### Using Bag of Words Vectorizer with Naive Bias

```python
bow_vectorizer = CountVectorizer()
```

```python
X_train_bow = bow_vectorizer.fit_transform(X_train)
```

```python
X_test_bow = bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
model_nb = MultinomialNB()
```

```python
model_nb.fit(X_train_bow , y_train)
```

```python
y_pred = model_nb.predict(X_test_bow)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test , y_pred)
```

### Using TF-ID Vectorizer with Naive Bias

```python
tf_vectorizer = TfidfVectorizer()
```

```python
X_train_tf = tf_vectorizer.fit_transform(X_train)
```

```python
X_test_tf = tf_vectorizer.transform(X_test)
```

```python
model_nb2 = MultinomialNB()
```

```python
model_nb2.fit(X_train_tf , y_train)
```

```python
y_pred_tf = model_nb2.predict(X_test_tf)
```

```python
accuracy_score(y_test , y_pred_tf)
```

### Using TF-ID Vectorizer with Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
model_LR = LogisticRegression()
```

```python
model_LR.fit(X_train_tf , y_train)
```

```python
y_pred = model_LR.predict(X_test_tf)
```

```python
accuracy_score(y_pred , y_test) #heighest accuracy
```

```python
import joblib

# Save the TF-IDF vectorizer
joblib.dump(tf_vectorizer, "NLP_Emotions_vectorizer.joblib")

# Save the trained Logistic Regression model
joblib.dump(model_LR, "NLP_Emotions_model.joblib")

print("Model and vectorizer saved successfully!")
print("Number of TF-IDF features:", len(tf_vectorizer.vocabulary_))
```
